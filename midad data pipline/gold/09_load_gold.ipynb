###**Gold Layer**
 Data Transformation & Integration

### Perform Joins

```sql
-- Combines all Silver sources into a unified Gold view.

CREATE OR REPLACE TEMP VIEW v_gold_all AS

-- Combines blog data.

SELECT
    content_id,
    title,
    description,
    content_type,
    topic,
    category,
    CAST(NULL AS STRING) AS difficulty_level,
    list_of_keywords,
    language,
    source,
    CAST(published_date AS TIMESTAMP) AS published_date,
    CAST(last_updated AS TIMESTAMP) AS last_updated,
    url
FROM edtech.silver.blogs

UNION ALL

-- Combines newsletter data.

SELECT
    content_id,
    title,
    description,
    content_type,
    topic,
    category,
    CAST(NULL AS STRING) AS difficulty_level,
    list_of_keywords,
    language,
    source,
    CAST(published_date AS TIMESTAMP) AS published_date,
    CAST(last_updated AS TIMESTAMP) AS last_updated,
    url
FROM edtech.silver.newsletters

UNION ALL

-- Combines Coursera course data.

SELECT
    content_id,
    title,
    description,
    content_type,
    topic,
    category,
    difficulty_level,
    list_of_keywords,
    language,
    source,
    CAST(published_date AS TIMESTAMP) AS published_date,
    CAST(last_updated AS TIMESTAMP) AS last_updated,
    url
FROM edtech.silver.coursera_courses

UNION ALL

-- Combines Microsoft Learn data.

SELECT
    content_id,
    title,
    description,
    content_type,
    topic,
    category,
    difficulty_level,
    list_of_keywords,
    language,
    source,
    CAST(published_date AS TIMESTAMP) AS published_date,
    CAST(last_updated AS TIMESTAMP) AS last_updated,
    url
FROM edtech.silver.microsoft_learn

UNION ALL

-- Combines GitHub repository data.

SELECT
    content_id,
    title,
    description,
    content_type,
    topic,
    category,
    difficulty_level,
    list_of_keywords,
    language,
    source,
    CAST(published_date AS TIMESTAMP) AS published_date,
    CAST(last_updated AS TIMESTAMP) AS last_updated,
    url
FROM edtech.silver.github_repos

UNION ALL

-- Combines YouTube video data.

SELECT
    content_id,
    title,
    description,
    content_type,
    topic,
    category,
    difficulty_level,
    list_of_keywords,
    language,
    source,
    CAST(published_date AS TIMESTAMP) AS published_date,
    CAST(last_updated AS TIMESTAMP) AS last_updated,
    url
FROM edtech.silver.youtube_videos;
```

```sql
-- Unifies tables, reorders fields, and saves the Gold data.

CREATE OR REPLACE TABLE edtech.gold.edtech_content AS

SELECT
    content_id,
    title,
    description,
    content_type,
    category,
    topic,
    difficulty_level,
    language,
    list_of_keywords,
    source,
    published_date,
    last_updated,
    url
FROM v_gold_all
ORDER BY published_date DESC NULLS LAST;

-- Sorts data by publication date.

SELECT *
FROM edtech.gold.edtech_content
ORDER BY published_date DESC NULLS LAST;
```

```sql
-- Verifies that the Gold table is registered in the catalog.

SHOW TABLES IN edtech.gold;

DESCRIBE TABLE edtech.gold.edtech_content;
```

```sql
%python
import base64
import pandas as pd

folder_path = "/Volumes/edtech/gold/exports/gold_content_snapshot/"
spark.table("edtech.gold.edtech_content").write.mode("overwrite").parquet(folder_path)

df = pd.read_parquet(folder_path)


single_file_path = "/tmp/gold_content_snapshot.parquet"
df.to_parquet(single_file_path)
print(f"Ready to download: {len(df)} rows")

with open(single_file_path, "rb") as f:
    file_bytes = f.read()
b64_data = base64.b64encode(file_bytes).decode()

html_button = f"""
<a download="gold_content_snapshot.parquet"
   href="data:application/octet-stream;base64,{b64_data}">
  <button style="padding:10px 20px; font-size:14px; background-color:#1B3139;
                 color:white; border:none; border-radius:4px; cursor:pointer;">
    ⬇️ Download gold_content_snapshot.parquet ({len(df)} rows)
  </button>
</a>
"""

displayHTML(html_button)
```
