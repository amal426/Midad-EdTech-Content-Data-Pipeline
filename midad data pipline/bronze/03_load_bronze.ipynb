**Loads raw API and RSS data from JSON files into Bronze tables in Databricks for storage and further processing**

```sql
-- List raw JSON files available in the Volume
LIST '/Volumes/edtech/bronze/raw_files/';
```

```sql
-- Ingest API JSON into api_raw
COPY INTO edtech.bronze.api_raw
FROM '/Volumes/edtech/bronze/raw_files/api_sources_1200.json'
FILEFORMAT = JSON
FORMAT_OPTIONS ('mergeSchema' = 'true', 'multiLine' = 'true')
COPY_OPTIONS ('mergeSchema' = 'true');
```

```sql
-- Ingest RSS JSON into rss_raw
COPY INTO edtech.bronze.rss_raw
FROM '/Volumes/edtech/bronze/raw_files/rss_feeds_600.json'
FILEFORMAT = JSON
FORMAT_OPTIONS ('mergeSchema' = 'true', 'multiLine' = 'true')
COPY_OPTIONS ('mergeSchema' = 'true');
```
