###**Bronze Layer** 

Creates two empty Bronze tables to store raw data from API and RSS sources

```sql
--  create Bronze tables with Column Mapping enabled
CREATE TABLE IF NOT EXISTS edtech.bronze.api_raw
USING DELTA
TBLPROPERTIES ('delta.columnMapping.mode' = 'name');

CREATE TABLE IF NOT EXISTS edtech.bronze.rss_raw
USING DELTA
TBLPROPERTIES ('delta.columnMapping.mode' = 'name');
```

###Verify Column Mapping

```sql
-- Verify that Column Mapping is enabled on both tables
SHOW TBLPROPERTIES edtech.bronze.api_raw;
SHOW TBLPROPERTIES edtech.bronze.rss_raw;
```
