###**create Schema for Medallion Architecture**
Sets up the Bronze, Silver, and Gold layers and creates a Volume for raw JSON files

```sql
-- create Schema for Medallion Architecture
CREATE CATALOG IF NOT EXISTS edtech;

CREATE SCHEMA IF NOT EXISTS edtech.bronze;
CREATE SCHEMA IF NOT EXISTS edtech.silver;
CREATE SCHEMA IF NOT EXISTS edtech.gold;
```

```sql
-- Create the Volume that stores raw JSON files
CREATE VOLUME IF NOT EXISTS edtech.bronze.raw_files;
```
