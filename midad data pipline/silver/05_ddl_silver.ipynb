###**Silver Layer**


Create a separate Silver table for each data source.


```sql
-- Create silver.blogs table schema
CREATE TABLE IF NOT EXISTS edtech.silver.blogs (
  content_id          STRING    NOT NULL,
  title               STRING    NOT NULL,
  description         STRING,
  topic               STRING,
  category            STRING,
  list_of_keywords    STRING,
  language            STRING,
  source              STRING,
  url                 STRING    NOT NULL,
  content_type        STRING,
  published_date      TIMESTAMP,
  last_updated        TIMESTAMP,
  silver_ingested_at  TIMESTAMP
)
USING DELTA;
```
